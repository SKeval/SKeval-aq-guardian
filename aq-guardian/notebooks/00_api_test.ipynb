```python
# UBA API test
uba_meta_url = "https://luftqualitaet.api.bund.dev/components/json"
r = requests.get(uba_meta_url)
print("UBA status:", r.status_code)
if r.status_code == 200:
    data = r.json()
    print("Keys:", list(data.keys())[:5])
else:
    print("UBA error")
```

```python
# Open-Meteo test (Berlin)
lat, lon = 52.52, 13.41
weather_url = (
    "https://api.open-meteo.com/v1/dwd-icon"
    f"?latitude={lat}&longitude={lon}"
    "&hourly=temperature_2m,relative_humidity_2m,wind_speed_10m,pressure_msl"
    "&past_days=1"
)
w = requests.get(weather_url)
print("Open-Meteo status:", w.status_code)
if w.status_code == 200:
    wdata = w.json()
    df = pd.DataFrame({
        "time": wdata["hourly"]["time"],
        "temperature": wdata["hourly"]["temperature_2m"],
    })
    print(df.head())
else:
    print("Weather error")
```
